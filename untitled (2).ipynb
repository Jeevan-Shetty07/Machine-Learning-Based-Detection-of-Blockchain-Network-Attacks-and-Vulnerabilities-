```python
from sklearn.preprocessing import LabelEncoder

le1 = LabelEncoder()
le2 = LabelEncoder()

df[' ERC20 most sent token type'] = le1.fit_transform(
    df[' ERC20 most sent token type']
)

df[' ERC20_most_rec_token_type'] = le2.fit_transform(
    df[' ERC20_most_rec_token_type']
)
```

```python
df.head()
```

```python
df.isnull().sum().sum()
```

```python
X = df.drop('FLAG', axis=1)

y = df['FLAG']
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled.shape
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)
```

```python
rf_acc = accuracy_score(y_test, rf_pred)

rf_precision = precision_score(y_test, rf_pred)

rf_recall = recall_score(y_test, rf_pred)

rf_f1 = f1_score(y_test, rf_pred)

print("Accuracy:", rf_acc)
print("Precision:", rf_precision)
print("Recall:", rf_recall)
print("F1 Score:", rf_f1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(6,4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from sklearn.svm import SVC

svm = SVC(
    kernel='rbf',
    random_state=42
)

svm.fit(X_train_scaled, y_train)

svm_pred = svm.predict(X_test_scaled)
```

```python
svm_acc = accuracy_score(y_test, svm_pred)

svm_precision = precision_score(y_test, svm_pred)

svm_recall = recall_score(y_test, svm_pred)

svm_f1 = f1_score(y_test, svm_pred)

print("Accuracy:", svm_acc)
print("Precision:", svm_precision)
print("Recall:", svm_recall)
print("F1 Score:", svm_f1)
```

```python
fraud_probability = xgb.predict_proba(X_test)

fraud_probability[:5]
```

```python
risk_prob = fraud_probability[:,1]

risk_prob[:10]
```

```python
risk_score = risk_prob * 100
```

```python
risk_score[:10]
```

```python
import pandas as pd

risk_category = []

for score in risk_score:

    if score <= 30:
        risk_category.append("Low Risk")

    elif score <= 70:
        risk_category.append("Medium Risk")

    else:
        risk_category.append("High Risk")
```

```python
risk_df = pd.DataFrame({
    'Actual_Class': y_test.values,
    'Fraud_Probability': risk_prob,
    'Risk_Score': risk_score,
    'Risk_Category': risk_category
})

risk_df.head()
```
